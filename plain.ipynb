```python
import pandas as pd
import json
```

```python
with open('project json/t20_wc_match_results.json') as f:
    data=json.load(f)
df_match=pd.DataFrame(data[0]['matchSummary'])
df_match.head()
```

```python
df_match.shape
```

```python
df_match.rename({'scorecard' : 'match_id'}, axis=1,inplace=True,)
df_match.head(5)
```

```python
match_id_dict={}
for index,row in df_match.iterrows():
    key1= row['team1'] +  '  Vs  '   + row['team2']
    key2= row['team2'] +  '  Vs    '    + row['team1']
    match_id_dict[key1]=row["match_id"]
    match_id_dict[key2]=row["match_id"]

match_id_dict
    
```

```python
df_match.to_csv('project csv/dim_match_summary.csv', index=False)
```

```python
with open('project json/t20_wc_batting_summary.json') as f:
    data=json.load(f)
    all_record=[]
    for rec in data:
        all_record.extend(rec['battingSummary'])
df_battin=pd.DataFrame(all_record)
df_battin.head(5)
df_battin.tail()
```

```python
df_battin["out/not_out"]=df_battin.dismissal.apply(lambda x:'out' if len(x)>0 else "not_out")
df_battin.head(11)
```

```python
df_battin["mach_id"] = df_battin["match"].map(match_id_dict)
df_battin.head()
```

```python
df_battin["mach_id"] = df_battin["match"].map(lambda x: match_id_dict.get(x, 'default_value'))
```

```python
df_battin.head()
```

```python
df_battin.to_csv('project csv/fact_batting_summary.csv',index=False)
```
